## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
df['is_declining_label'] = (df['trend_direction'] == 'down').astype(int)
print(df['is_declining_label'].value_counts(normalize=True))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
weekly_capacity = 50
total_candidates = len(df)
print(f"Reviewer capacity ~{weekly_capacity}/week vs {total_candidates} total pages")
print("→ Precision@K (top of queue) is the metric that matches the real decision, not overall accuracy.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
corr = df[['word_count', 'content_age_days', 'days_since_last_update', 'ctr', 'avg_position']].corrwith(df['is_declining_label'])
print(corr)
```
